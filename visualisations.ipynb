```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import requests
```

```python
from cleaning import cleaning_df #importing function from .py script

df=pd.read_csv("listening-history-2026-11-09.csv") #reading in the dataframe
st = cleaning_df(df) #cleaning the dataframe, st for storytel

st.columns
```

```python
from cloudcoverage import cloudy #importing function from .py script
start_cloud_values = [cloudy(date) for date in st.start.dt.date] #making a list of cloudcoverage on the startdates
st["start_clouds"] = start_cloud_values #adding cloudcoverage as a column to st
end_cloud_values = [cloudy(date) for date in st.stop.dt.date] #making a list of cloudcoverage on the startdates
st["end_clouds"] = end_cloud_values #adding cloudcoverage as a column to st
```

```python
fig, ax = plt.subplots()
ax.hist([st.start_clouds, st.end_clouds], )
```
